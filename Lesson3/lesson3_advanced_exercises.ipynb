## Exercise 2: Comparison chain

- Given `a = 12`, `b = 12`, `c = 7`, print at least 6 comparison results.
- Include equality, inequality, greater/less, and a chained comparison (e.g., `c < a <= b`).

```python
# Write your solution here
a = 12
b = 12
c = 7

print(a == b)       # Equality
print(a != c)       # Inequality
print(a > c)        # Greater than
print(c < a)        # Less than
print(a >= b)       # Greater than or equal to
print(c <= b)       # Less than or equal to
print(c < a <= b)   # Chained comparison
```

## Exercise 3: Operator precedence check

- Create a complex boolean expression using `and`, `or`, and `not`.
- Evaluate it once without extra parentheses and once with explicit parentheses.
- Print both and comment whether they match.

```python
# Write your solution here
p = True
q = False
r = True

without_parentheses = p or q and not r
with_parentheses = p or (q and (not r))

print(without_parentheses)
print(with_parentheses)

# Both results match because `not` is evaluated before `and`, and `and` before `or`.
print(without_parentheses == with_parentheses)
```

## Exercise 4: Identity vs equality

- Create two equal lists and compare with `==` and `is`.
- Assign one list to another variable and compare again.
- Write a short comment explaining differences in output.

```python
# Write your solution here
list1 = [1, 2, 3]
list2 = [1, 2, 3]

print(list1 == list2)
print(list1 is list2)

list3 = list1

print(list1 == list3)
print(list1 is list3)

# `==` compares values, while `is` compares object identity.
# list1 and list2 have equal contents but are different objects.
# list1 and list3 refer to the same object.
```

## Exercise 5: Membership filter

- Given `items = ["python", "java", "go", "rust"]`, test membership for 4 queries.
- Use both `in` and `not in`.
- Print human-readable results like `python: found`.

```python
# Write your solution here
items = ["python", "java", "go", "rust"]

print(f"python: {'found' if 'python' in items else 'not found'}")
print(f"java: {'found' if 'java' in items else 'not found'}")
print(f"javascript: {'not found' if 'javascript' not in items else 'found'}")
print(f"c++: {'not found' if 'c++' not in items else 'found'}")
```
